### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
# [Paste your code here]
# --- Budget Tracker Program ---

# Ask for the user's name
name = input("What is your name? ")

# Ask for monthly income
monthly_income = float(input("What is your monthly income? "))

print("\nEnter up to 3 monthly expenses.")
print("If you have fewer than 3 expenses, enter 0 for the remaining ones.\n")

# Ask for up to 3 expenses
expense1 = float(input("Expense 1: "))
expense2 = float(input("Expense 2: "))
expense3 = float(input("Expense 3: "))

# Calculate total expenses
total_expenses = expense1 + expense2 + expense3

# Calculate leftover money
leftover = monthly_income - total_expenses

# Print results
print("\n--- Budget Summary ---")
print(f"Hello {name}, here is your monthly budget breakdown:")
print(f"Total Income: ${monthly_income}")
print(f"Total Expenses: ${total_expenses}")
print(f"Money Left Over: ${leftover}")

# Warn the user if the budget goes negative
if leftover < 0:
    print("\n⚠ Warning: Your budget is negative! You are overspending.")
else:
    print("\nGreat job! You are staying within your budget.")
```
