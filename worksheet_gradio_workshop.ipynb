## A quick reminder about objects

Before we bring in the interface, a short recap. If any of this feels shaky, the **Python OOP Quick Reference** handout covers it in more detail.

- A **class** is a template. It describes what something should look like.
- An **object** is one individual item made from that template.
- Each object **stores information** (its attributes) and **can do things** (its methods).

In our finance examples, `Transaction` is the template. Each actual transaction (your coffee, your salary) is an object made from it. Each one remembers its own description and amount, and can perform small actions like describing itself.

Here is the idea that ties today together: **everything you touch in Gradio is also an object.** A text box is an object. A button is an object. So building an interface is not a new skill. It is just more objects, talking to the objects you already know how to write.

## See it running

Run the two cells below. The first installs Gradio. The second builds a tiny app and launches it. A small interface will appear underneath the cell.

Add a transaction, then read the code and match what you see on screen to what is written. Watch the flow: the interface objects collect what you type, hand it to a function, the function builds a `Transaction` object, and hands a friendly message back to the screen.

```python
# Install Gradio (run this first)
!pip install -q gradio
```

```python
import gradio as gr
from datetime import datetime


class Transaction:
    """A blueprint for a single financial transaction."""

    def __init__(self, description: str, amount: float):
        # These are the attributes each transaction object remembers.
        self.description = description
        self.amount = amount
        self.timestamp = datetime.now()

    def __repr__(self) -> str:
        # A method: a clean, printable version of the object.
        return f"Transaction('{self.description}', ${self.amount:.2f})"


def add_transaction(description, amount):
    """The glue: takes what the interface collected, builds an object, returns a message."""
    new_transaction = Transaction(description=description, amount=amount)
    print(f"Object created: {new_transaction}")  # prints to the Colab log
    return f"Added: {new_transaction.description} for ${new_transaction.amount:.2f}"


# The interface. Each line here creates a Gradio object.
with gr.Blocks() as demo:
    gr.Markdown("# Simple Transaction Logger")
    gr.Markdown("A worked example of interface objects talking to a Transaction object.")

    input_desc = gr.Textbox(label="Description", placeholder="e.g. Coffee")
    input_amount = gr.Number(label="Amount", value=0.0)
    add_button = gr.Button(value="Add Transaction")
    output_message = gr.Textbox(label="Confirmation", interactive=False)

    # Tell the button object what to do when it is clicked.
    add_button.click(
        fn=add_transaction,
        inputs=[input_desc, input_amount],
        outputs=output_message,
    )

demo.launch(debug=True)
```

## What just happened

Four things, in order:

1. The **interface objects** (`gr.Textbox`, `gr.Number`) collected what you typed.
2. Clicking the button called the `add_transaction` function and passed those values in.
3. The function built a **`Transaction` object** from your values, exactly the way you would in a plain notebook.
4. The function returned a string, which the **output object** displayed on screen.

That is the whole pattern. The interface never needs to know how `Transaction` works inside. It just collects input, hands it over, and shows what comes back. Keeping the interface and your business logic separate like this is what makes the app easy to extend later.

```python
class SavingsGoal:
    def __init__(self, goal_name, target_amount, current_savings, timeframe_months, priority_level):
        self.goal_name = goal_name
        self.target_amount = target_amount
        self.current_savings = current_savings
        self.timeframe_months = timeframe_months
        self.priority_level = priority_level

    def monthly_required(self):
        remaining = self.target_amount - self.current_savings
        if self.timeframe_months > 0:
            return remaining / self.timeframe_months
        return None

    def progress_percentage(self):
        return (self.current_savings / self.target_amount) * 100

    def summary(self):
        monthly = self.monthly_required()
        progress = self.progress_percentage()
        return (
            f"Goal: {self.goal_name}\n"
            f"Priority: {self.priority_level}\n"
            f"Progress: {progress:.1f}%\n"
            f"Monthly required: ${monthly:.2f}"
        )
```

```python
import gradio as gr

def build_goal(goal_name, target_amount, current_savings, timeframe_months, priority_level):
    goal = SavingsGoal(goal_name, target_amount, current_savings, timeframe_months, priority_level)
    return goal.summary()

demo = gr.Interface(
    fn=build_goal,
    inputs=[
        gr.Textbox(label="Goal Name"),
        gr.Number(label="Target Amount ($)"),
        gr.Number(label="Current Savings ($)"),
        gr.Number(label="Timeframe (months)"),
        gr.Dropdown(["High", "Medium", "Low"], label="Priority Level")
    ],
    outputs=gr.Textbox(label="Savings Goal Summary")
)

demo.launch()
```
