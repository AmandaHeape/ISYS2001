## Part 1: Setup and expense entry

Begin with the user setup, then allow the user to enter as many expenses as they like in one session.

The budget validation below is provided as a worked pattern. Your task is the expense entry loop beneath it. Keep asking for expenses until the user types `done`, and for each valid amount, store it, add it to the running total, and give brief feedback.


```python
print("=== Loop-Powered Finance Tracker ===")
print()

name = input("Enter your name: ")

# Budget validation, provided as a worked pattern.
# Keeps asking until the number falls within a sensible range.
monthly_budget = 0
while monthly_budget < 100 or monthly_budget > 10000:
    try:
        budget_input = input("Enter monthly budget ($100 to $10,000): $")
        monthly_budget = float(budget_input)
        if monthly_budget < 100:
            print("Budget too low. Minimum is $100.")
        elif monthly_budget > 10000:
            print("Budget seems high. Maximum is $10,000.")
    except ValueError:
        print("Please enter a valid number.")

print(f"\nWelcome {name}. Monthly budget: ${monthly_budget:.2f}")

# ===== Your task: multi-expense entry =====
expenses = []      # every amount is stored here
total_spent = 0

print("\n--- Enter your expenses ---")
print("Type 'done' when you are finished.")

while True:
    entry = input("\nExpense amount (or 'done'): $")

    if entry.lower() == "done":
        break
    try:
        amount = float(entry)
        if amount <= 0:
            print("Amount must be positive.")
            continue
    except ValueError:
        print("Please enter a valid number.")
        continue

    expenses.append(amount)
    total_spent += amount
    print(f"Added ${amount:.2f}. Running total: ${total_spent:.2f}")

    # Your task:
    # 1. try to convert entry into a float
    # 2. if it is not a positive number, tell the user and continue
    # 3. if it is valid, append it to expenses, add it to total_spent,
    #    and print the amount along with the running total


print(f"\nEntry complete: {len(expenses)} expenses, ${total_spent:.2f} total")
```

## Part 2: Summarise the numbers

You now have a list of expenses, so the next step is to summarise it. The total and average are provided. Your task is to find the highest and lowest values, and to count the large expenses.


```python
if len(expenses) > 0:
    print("\n--- Analysis ---")
    average = total_spent / len(expenses)
    print(f"Total spent: ${total_spent:.2f}")
    print(f"Average expense: ${average:.2f}")
    print(f"Budget used: {(total_spent / monthly_budget) * 100:.1f}%")

    # Your task: find the highest and lowest expense.
    # Hint: start both at the first item, then loop and compare.
    highest = expenses[0]
    lowest = expenses[0]

    #loop to find real highest & lowest
    for amount in expenses:
      if amount > highest:
        highest = amount
      if amount < lowest:
        lowest = amount


    print(f"Highest expense: ${highest:.2f}")
    print(f"Lowest expense: ${lowest:.2f}")

    # Your task: count how many expenses were more than 20% of the budget.
    large_count = 0
    #expenses over 20% of budget
    for amount in expenses:
      if amount > monthly_budget * 0.2:
        large_count += 1


    print(f"Large expenses (over 20% of budget): {large_count}")

else:
    print("\nNo expenses entered, so there is nothing to analyse.")
```

## Part 3: A menu that keeps running

Wrap the program in a menu so the user can keep working until they choose to quit. Option 4 (quit) is provided. Complete options 1, 2 and 3 using patterns from earlier in the notebook.


```python
running = True

while running:
    print("\n" + "-" * 30)
    print(f"{name}'s finance tracker")
    print("-" * 30)
    print("1. View all expenses")
    print("2. View summary")
    print("3. Add an expense")
    print("4. Quit")

    choice = input("\nChoose an option (1-4): ")

    if choice == "1":
        # Your task: print each expense with a number in front of it,
        # for example "1. $23.50". Handle the empty-list case as well.
        if len(expenses) == 0:
          print("No expenses are recorded yet.")
        else:
          for i in range(len(expenses)):
            print(f"{i+1}. ${expenses[i]:.2f}")

    elif choice == "2":
        # Your task: print the total and the number of expenses.
        if len(expenses) == 0:
          print("No expenses")
        else:
          total = sum(expenses)
          print(f"Total spend: ${total:.2f}")
          print(f"number of expense: {len(expenses)}")

    elif choice == "3":
        # Your task: ask for one more expense, validate it, and add it
        # to the list and the total. You can reuse your Part 1 logic.
        while True:
          entry = input("enter amount: $")
          try:
            amount = float(entry)
            if amount <= 0:
              print("amount must be positive")
              continue
            break
          except ValueError:
            print("please enter a valid number")

        expenses.append(amount)
        print(f"Added ${amount:.2f}.")


    elif choice == "4":
        running = False
        print(f"\nGoodbye, {name}.")

    else:
        print("Not a valid choice. Please pick 1 to 4.")
```

## Stretch (optional)

If you finish early, add a savings goals section. Ask how many goals the user wants to track (1 to 5), then use a `for` loop to collect a name and target for each, and report how much remains to be saved. This section is not assessed.


```python
running = True

while running:
    print("\n" + "-" * 30)
    print(f"{name}'s finance tracker")
    print("-" * 30)
    print("1. View all expenses")
    print("2. View summary")
    print("3. Add an expense")
    print("4. Track savings goals")
    print("5. Quit")


    choice = input("\nChoose an option (1-4): ")

    if choice == "1":
        # Your task: print each expense with a number in front of it,
        # for example "1. $23.50". Handle the empty-list case as well.
        if len(expenses) == 0:
          print("No expenses are recorded yet.")
        else:
          for i in range(len(expenses)):
            print(f"{i+1}. ${expenses[i]:.2f}")

    elif choice == "2":
        # Your task: print the total and the number of expenses.
        if len(expenses) == 0:
          print("No expenses")
        else:
          total = sum(expenses)
          print(f"Total spend: ${total:.2f}")
          print(f"number of expense: {len(expenses)}")

    elif choice == "3":
        # Your task: ask for one more expense, validate it, and add it
        # to the list and the total. You can reuse your Part 1 logic.
        while True:
          entry = input("enter amount: $")
          try:
            amount = float(entry)
            if amount <= 0:
              print("amount must be positive")
              continue
            break
          except ValueError:
            print("please enter a valid number")

        expenses.append(amount)
        print(f"Added ${amount:.2f}.")

    elif choice == "4":
    # Ask how many goals (1–5), with validation
      while True:
        try:
            num_goals = int(input("How many goals do you want to track? (1–5): "))
            if 1 <= num_goals <= 5:
                break
            else:
                print("Please enter a number between 1 and 5.")
        except ValueError:
            print("Please enter a valid whole number.")

    # For each goal, ask name, target, current, then show remaining
      for i in range(num_goals):
          print(f"\nGoal {i+1}:")
          name = input("Name of goal: ")

          # Target amount
          while True:
              try:
                  target = float(input("Target amount: $"))
                  if target > 0:
                      break
                  else:
                      print("Target must be positive.")
              except ValueError:
                  print("Please enter a valid number.")

          # Current savings
          while True:
              try:
                  current = float(input("Current saved: $"))
                  if current >= 0:
                      break
                  else:
                      print("Current savings cannot be negative.")
              except ValueError:
                  print("Please enter a valid number.")

          remaining = target - current
          if remaining <= 0:
              print(f"{name}: goal reached!")
          else:
              print(f"{name}: ${remaining:.2f} left to save.")

    elif choice == "5":
      running = False
      print(f"\nGoodbye, {name}.")

    else:
        print("Not a valid choice. Please pick 1 to 4.")
```
