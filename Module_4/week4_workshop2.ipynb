## Part 1: From variables to collections

Here is the problem with the Workshop 1 approach: it keeps a running total but throws away the individual expenses, so afterwards there is nothing left to analyse.


```python
print("=== The limitation ===")

total = 0
count = 0

for i in range(3):
    expense = (i + 1) * 25  # 25, 50, 75
    total += expense
    count += 1
    print(f"Expense {i + 1}: ${expense}")

print(f"\nTotal: ${total}, count: {count}")
print("The individual amounts are gone. We cannot find the largest,")
print("categorise them, or search them later.")
```

My notes:
What the for i in range(3) loop is doing

range(3) produces the sequence: 0,1,2

So the loop runs three times, with i taking each of those values in order.

Inside the loop:
expense = (i + 1) * 25

Because i is 0, 1, 2:

When i = 0 → (0 + 1) * 25 = 25

When i = 1 → (1 + 1) * 25 = 50

When i = 2 → (2 + 1) * 25 = 75

So the loop generates three fixed expenses: 25, 50, 75.

Then each loop repetition:

adds the expense to total

increases count

prints the expense

This is a count‑controlled loop:
You already know exactly how many expenses you will process (three), so a for loop is the correct pattern.

Why the limitation happens
After the loop finishes, you only have:
total
count

You did not store the individual expenses anywhere (like in a list).
They were printed, added to the total, and then lost.

That’s why the program prints:
The individual amounts are gone. We cannot find the largest,
categorise them, or search them later.

Because without storing them, you cannot:

find the largest expense

check which ones are above a threshold

categorise them

search through them

print them again later

A for loop alone is not enough — you need a list if you want to keep the values.

Storing the amounts in a list fixes this. The same collection can be analysed in as many ways as you like, after the fact.


```python
print("=== The collection approach ===")

expenses = []
total = 0

for i in range(3):
    expense = (i + 1) * 25  # 25, 50, 75
    expenses.append(expense)
    total += expense

print(f"All expenses: {expenses}")
print(f"Largest: ${max(expenses)}")
print(f"Smallest: ${min(expenses)}")
print(f"Average: ${total / len(expenses):.2f}")
```

The point: a variable holds one value, while a list holds many values in order under a single name. Once the data is in a list, analysis becomes possible.


## Part 2: List indexing and safe access

Lists use numbered positions starting at 0. This is the source of a lot of early errors, so it is worth seeing clearly.


```python
monthly_expenses = [450.00, 380.50, 520.25, 290.75]

print(f"List: {monthly_expenses}")
print(f"Positions:  0       1       2       3")
print()
print(f"First (index 0): ${monthly_expenses[0]}")
print(f"Second (index 1): ${monthly_expenses[1]}")
print(f"Last (index -1): ${monthly_expenses[-1]}")
print(f"Number of items: {len(monthly_expenses)}")

# A common mistake: reaching past the end of the list
try:
    print(monthly_expenses[4])
except IndexError as e:
    print(f"\nIndexError: {e}")
    print("Four items have indices 0, 1, 2, 3, not 1 to 4.")
```

Before reading from a list, check it actually has items in it. This avoids a crash on an empty list.


```python
transactions = [25.50, 67.25, 15.00]

print("=== Safe transaction display ===")

if len(transactions) > 0:
    print(f"Total transactions: {len(transactions)}")
    for i in range(len(transactions)):
        print(f"{i + 1}. ${transactions[i]:.2f}")
else:
    print("No transactions recorded yet.")

# The same check keeps an empty list safe
empty = []
if len(empty) > 0:
    print(empty[0])
else:
    print("\nEmpty list handled safely.")
```

## Part 3: Collection processing patterns

Three patterns cover most analysis of a collection: accumulation (sum or average), conditional counting, and finding extremes. They work on any list, not just expenses.


```python
expenses = [23.50, 156.75, 45.00, 89.25, 12.80, 234.60, 67.30]

# Pattern 1: accumulation
total = 0
for expense in expenses:
    total += expense
average = total / len(expenses)
print(f"Total: ${total:.2f}")
print(f"Average: ${average:.2f}")

# Pattern 2: conditional counting
threshold = 100
large = 0
for expense in expenses:
    if expense > threshold:
        large += 1
print(f"Expenses over ${threshold}: {large}")

# Pattern 3: finding extremes
highest = expenses[0]
lowest = expenses[0]
for expense in expenses:
    if expense > highest:
        highest = expense
    if expense < lowest:
        lowest = expense
print(f"Highest: ${highest:.2f}")
print(f"Lowest: ${lowest:.2f}")
```

Analysing by category needs two lists that line up by position: `expenses[i]` belongs to `categories[i]`. Index-based looping lets you match them.


```python
expenses = [23.50, 156.75, 45.00, 89.25, 12.80, 234.60, 67.30]
categories = ["Food", "Rent", "Food", "Gas", "Food", "Rent", "Entertainment"]

food_count = 0
food_total = 0
rent_count = 0
rent_total = 0

for i in range(len(expenses)):
    if categories[i] == "Food":
        food_count += 1
        food_total += expenses[i]
    elif categories[i] == "Rent":
        rent_count += 1
        rent_total += expenses[i]

print("=== Category summary ===")
print(f"Food: {food_count} transactions, ${food_total:.2f} total")
print(f"Rent: {rent_count} transactions, ${rent_total:.2f} total")
```

*The core issue*

Right now you keep amounts in one list and categories in another list:

This works, but only if both lists stay perfectly aligned.
If one list changes, everything breaks.

**Clearer ways to keep an amount with its category**

There are two common alternatives:



1.   **A list of pairs (tuples or lists)**

Example idea:
[ [23.50, "Food"], [156.75, "Rent"], ... ]
**Why it’s clearer:**


*   Each transaction is stored as **one unit**
*   You never worry about matching indexes
*   Looping becomes more readable: “for each transaction”

**Trade‑offs:**


*   Slightly more complex structure for beginners
*   You must unpack two values inside the loop


*   Still simple enough for your workshop level


2.   **A list of dictionaries**

Example idea:
{ "amount": 23.50, "category": "Food" }

**Why it’s clearer:**
*  Each transaction has named fields

*  You don’t rely on remembering index positions

*  Much easier to extend later (date, notes, tags)

**Trade‑offs:**

*  More typing

*  More advanced concept (your workshop may not want this yet)

*  Slightly slower to write but much clearer to read

**Why your current method is harder**

Using two separate lists means:

*  You must keep them the same length

*  You must always use i to match them

*  You can accidentally misalign them

*  It’s harder to add new attributes later (e.g., merchant, date)

It’s not wrong — it’s just fragile.

**Simple takeaway**

Keeping each transaction together (as a pair or dictionary) makes your code:

*  safer

*  easier to read

*  easier to extend

*  less dependent on matching indexes

Your current method is fine for learning loops, but combining the data makes the program more robust once you move beyond beginner exercises.

## Part 4: Search and filter

Searching a collection is the same shape as analysis: loop through, test each item, and act on the ones that match.


```python
amounts = [23.50, 156.75, 45.00, 89.25, 12.80, 234.60, 67.30]
descriptions = ["Grocery", "Rent", "Restaurant", "Gas", "Coffee", "Rent", "Movie"]
dates = ["Jan 1", "Jan 1", "Jan 2", "Jan 3", "Jan 3", "Feb 1", "Feb 5"]

# Search by description
search_term = "Rent"
matches = 0
print(f"Transactions matching '{search_term}':")
for i in range(len(descriptions)):
    if search_term.lower() in descriptions[i].lower():
        matches += 1
        print(f"  {dates[i]}: ${amounts[i]:.2f} - {descriptions[i]}")
print(f"Found {matches} match(es).")
```

Filtering by an amount range works the same way, testing each amount against a lower and upper bound.


```python
amounts = [23.50, 156.75, 45.00, 89.25, 12.80, 234.60, 67.30]
descriptions = ["Grocery", "Rent", "Restaurant", "Gas", "Coffee", "Rent", "Movie"]

low = 40
high = 100

matching_count = 0
matching_total = 0
print(f"Transactions between ${low} and ${high}:")
for i in range(len(amounts)):
    if low <= amounts[i] <= high:
        matching_count += 1
        matching_total += amounts[i]
        print(f"  ${amounts[i]:.2f} - {descriptions[i]}")

print(f"\nMatching: {matching_count}, totalling ${matching_total:.2f}")
```

Two conditions can be combined with `and`. Here we find January transactions over $20.


```python
amounts = [23.50, 156.75, 45.00, 89.25, 12.80, 234.60, 67.30]
descriptions = ["Grocery", "Rent", "Restaurant", "Gas", "Coffee", "Rent", "Movie"]
dates = ["Jan 1", "Jan 1", "Jan 2", "Jan 3", "Jan 3", "Feb 1", "Feb 5"]

print("January transactions over $20:")
for i in range(len(amounts)):
    if "Jan" in dates[i] and amounts[i] > 20:
        print(f"  {dates[i]}: ${amounts[i]:.2f} - {descriptions[i]}")
```

## Part 5: An interactive menu

Everything so far combines into a tracker that keeps running until the user quits. Each menu option reuses a pattern you have already seen: validation to add, safe access to view, search to find, and the analysis patterns to summarise.


```python
print("=== Interactive finance tracker ===")

name = input("Enter your name: ")

budget = 0
while budget < 100 or budget > 50000:
    try:
        budget = float(input("Monthly budget ($100 to $50,000): $"))
        if budget < 100:
            print("Budget too low. Minimum is $100.")
        elif budget > 50000:
            print("Budget too high. Maximum is $50,000.")
    except ValueError:
        print("Please enter a valid number.")
        budget = 0

print(f"\nWelcome {name}. Budget: ${budget:.2f}")

amounts = []
descriptions = []
running = True

while running:
    print("\n" + "=" * 34)
    print(f"{name}'s tracker")
    print("=" * 34)
    print("1. Add an expense")
    print("2. View all expenses")
    print("3. Search by description")
    print("4. Analysis")
    print("5. Quit")

    choice = input("\nChoose an option (1-5): ")

    if choice == "1":
        amount = -1
        while amount < 0:
            try:
                amount = float(input("Amount: $"))
                if amount < 0:
                    print("The amount cannot be negative.")
            except ValueError:
                print("Please enter a valid number.")
                amount = -1
        description = input("Description: ")
        amounts.append(amount)
        descriptions.append(description)
        print(f"Added ${amount:.2f} for {description}.")

    elif choice == "2":
        if len(amounts) > 0:
            for i in range(len(amounts)):
                print(f"{i + 1}. ${amounts[i]:.2f} - {descriptions[i]}")
        else:
            print("No expenses recorded yet.")

    elif choice == "3":
        term = input("Search for: ")
        found = 0
        for i in range(len(descriptions)):
            if term.lower() in descriptions[i].lower():
                found += 1
                print(f"  ${amounts[i]:.2f} - {descriptions[i]}")
        print(f"Found {found} match(es).")

    elif choice == "4":
        if len(amounts) > 0:
            total = 0
            for amount in amounts:
                total += amount
            large = 0
            for amount in amounts:
                if amount > budget * 0.1:
                    large += 1
            print(f"Total spent: ${total:.2f}")
            print(f"Budget used: {(total / budget) * 100:.1f}%")
            print(f"Largest: ${max(amounts):.2f}")
            print(f"Large expenses (over 10% of budget): {large}")
        else:
            print("Nothing to analyse yet.")

    elif choice == "5":
        running = False
        print(f"Goodbye, {name}.")

    else:
        print("Not a valid choice. Please pick 1 to 5.")
```
