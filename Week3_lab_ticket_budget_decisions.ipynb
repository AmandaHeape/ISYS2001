## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Create a savings goal tracker in Python.
The user enters their savings goal and their current savings.
The program should calculate progress as a percentage and give different feedback depending on how close they are to the goal.
```

````python
# Paste the code your prompt produced here, then run it to show it works.
print("=== Savings Goal Tracker ===")

goal = float(input("Enter your savings goal: "))
current = float(input("Enter your current savings: "))
weekly_saving = float(input("How much do you save per week? "))

# Progress calculation
progress = (current / goal) * 100
print(f"Progress: {progress:.1f}%")

# Time until goal
remaining = goal - current

if remaining <= 0:
    weeks_left = 0
else:
    weeks_left = remaining / weekly_saving

# Feedback based on progress
if progress < 25:
    print("You're just getting started — keep going!")
elif progress < 75:
    print("Great progress — stay consistent!")
elif progress < 100:
    print("Almost there — you're so close!")
else:
    print("Goal achieved! Amazing work!")

# Time feedback
if weeks_left == 0:
    print("You've already reached your goal!")
else:
    print(f"Estimated time until goal: {weeks_left:.1f} weeks")
````
