Now that you have done the thinking, this is where AI earns its keep. Notice the difference:

Try your version with your preferred AI tool, then compare its code to your logic. Did it match your thinking? What did it do differently?

````python
# Implement your academic standing system here
def classify_cwa(cwa):
    if cwa >= 85:
        return "High Distinction"
    elif cwa >= 80:
        return "Distinction"
    elif cwa >= 65:
        return "Credit"
    elif cwa >= 50:
        return "Pass"
    else:
        return "Fail"


def academic_standing(cwa, failed_units, consecutive_fail_periods):
    if cwa >= 50 and failed_units == 0:
        return "Good Standing"

    if cwa < 50 or failed_units >= 1:
        if consecutive_fail_periods >= 3:
            return "Terminated"
        if consecutive_fail_periods >= 2:
            return "At Risk"
        return "Conditional"

    return "Good Standing"


def awards(swa, credits_this_period, swa_previous, credits_previous, cwa):
    if swa >= 85 and credits_this_period >= 75:
        return "Vice-Chancellor’s List"

    if swa >= 85 and swa_previous >= 85 and (credits_this_period + credits_previous) >= 75:
        return "Vice-Chancellor’s List"

    if cwa >= 80:
        return "Merit Award / Letter of Commendation"

    return "No Award"


def percentile_band(cwa):
    if cwa >= 85:
        return "Top 10%"
    elif cwa >= 75:
        return "Top 25%"
    elif cwa >= 65:
        return "Top 50%"
    elif cwa >= 50:
        return "Lower 50%"
    else:
        return "Bottom 10%"

def print_student_report(name, cwa, failed_units, consecutive_fail_periods,
                         swa, credits_this_period, swa_previous, credits_previous):

    print(f"--- {name} ---")
    print(f"CWA: {cwa}")
    print(f"Classification: {classify_cwa(cwa)}")
    print(f"Academic Standing: {academic_standing(cwa, failed_units, consecutive_fail_periods)}")
    print(f"Awards: {awards(swa, credits_this_period, swa_previous, credits_previous, cwa)}")
    print(f"Percentile Band: {percentile_band(cwa)}")
    print()  # blank line for spacing


# Example students
print_student_report(
    name="Student A",
    cwa=81,
    failed_units=0,
    consecutive_fail_periods=0,
    swa=81,
    credits_this_period=75,
    swa_previous=0,
    credits_previous=0
)

print_student_report(
    name="Student B",
    cwa=49,
    failed_units=2,
    consecutive_fail_periods=2,
    swa=49,
    credits_this_period=75,
    swa_previous=0,
    credits_previous=0
)

print_student_report(
    name="Student C",
    cwa=60,
    failed_units=0,
    consecutive_fail_periods=0,
    swa=60,
    credits_this_period=75,
    swa_previous=0,
    credits_previous=0
)

````

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

**Working with AI**
- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
- Give it your intent when you ask for help, especially when debugging.
