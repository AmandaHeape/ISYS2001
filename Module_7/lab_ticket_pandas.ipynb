## Task 1: total sales by region, highest first

Load the file into a DataFrame, clean whatever needs cleaning so the sale values are
real numbers, and produce the total `Total_Sale` for each `Region`, ordered from
highest to lowest.

Plan it in plain words in the cell below before you write the pandas. If you use an

My plan, in plain words:

*(write your plan here before the code)*

```python
# Task 1: total sales by region, highest first.

#Remove the dollar signs from Quantity, Unit_Price, and Total_Sale.

#Convert Quantity to an integer, and Unit_Price + Total_Sale to floats.

#Group the data by Region and sum the Total_Sale values.

#Sort the regions from highest total sales to lowest.

#Print the result.
```

```python
import pandas as pd

# 1. Load the file
df = pd.read_csv("sales_data.csv")

# 2. Clean columns with dollar signs
cols_to_clean = ["Quantity", "Unit_Price", "Total_Sale"]

for col in cols_to_clean:
    df[col] = df[col].str.replace("$", "", regex=False)

# 3. Convert to correct numeric types
df["Quantity"] = df["Quantity"].astype(int)
df["Unit_Price"] = df["Unit_Price"].astype(float)
df["Total_Sale"] = df["Total_Sale"].astype(float)

# 4. Group by Region and sum Total_Sale
region_totals = df.groupby("Region")["Total_Sale"].sum()

# 5. Sort highest → lowest
region_totals = region_totals.sort_values(ascending=False)

# 6. Display
region_totals
```

```python
# Task 3: your own question.
# already cleaned and Total_Sale is a float

# Group by Region and compute the average sale value
avg_sale_by_region = (
    df.groupby("Region")["Total_Sale"]
      .mean()
      .sort_values(ascending=False)
)

avg_sale_by_region

```
