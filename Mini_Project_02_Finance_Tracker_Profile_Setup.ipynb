## Step 5: Write the Code

**What to do:** Now turn your pseudocode into real Python, using only what you know so far: `input()`, `print()`, variables, and basic operations. Below is one example solution. Run it, read it line by line, and check how each part lines up with your pseudocode.

````python
# Collect user information for their financial profile
user_name = input("What is your name? ")
theme_color = input("What color would you like for your app theme? (e.g., blue, green, dark) ")
primary_currency = input("What is your primary currency? (e.g., AUD, USD, EUR) ")
main_savings_goal = input("What is one big savings goal you have? (e.g., New Car, Holiday) ")

# Construct the personalized welcome message
summary = ("Welcome to your Personal Finance Tracker, " + user_name + "! Your profile is ready. " + \
           "The app will use a " + theme_color + " theme and all amounts will be in " + \
           primary_currency + ". Let's start working towards your goal of saving for a " + \
           main_savings_goal + "!")

# Display the summary message
print(summary)
````

````python
# Step 5: Finance Tracker Profile Setup

# 1. Ask the user for their profile details
user_name = input("What is your name? ")
theme_color = input("What app theme colour would you like to use? ")
primary_currency = input("What is your primary currency (e.g., AUD, USD)? ")
income_frequency = input("How often do you receive income (e.g., weekly, fortnightly, monthly)? ")
monthly_budget_limit = input("What is your monthly budget limit? ")
main_savings_goal = input("What is your main savings goal (e.g., holiday, car, home deposit)? ")

# 2. Build the personalised welcome message
welcome_message = (
    f"Welcome to your Personal Finance Tracker, {user_name}! "
    f"Your profile is now set up. The app will use a {theme_color} theme, "
    f"track all amounts in {primary_currency}, and organise your income on a {income_frequency} schedule. "
    f"Your monthly budget limit is set to {monthly_budget_limit}. "
    f"Let’s start working towards your goal of saving for {main_savings_goal}!"
)

# 3. Print the final message
print("\n--- Finance Tracker Profile Summary ---")
print(welcome_message)
````

````python
# Your extended version.
# Start from the example above, then add at least one new feature of your own.
# Keep to what you know: input(), print(), variables, and basic operations.

# Your code here:

# Step 5: Finance Tracker Profile Setup (Extended Version)

# Collect user profile details
user_name = input("What is your name? ")
theme_color = input("What app theme colour would you like to use? ")
primary_currency = input("What is your primary currency (e.g., AUD, USD)? ")
income_frequency = input("How often do you receive income (e.g., weekly, fortnightly, monthly)? ")
monthly_budget_limit = input("What is your monthly budget limit? ")
main_savings_goal = input("What is your main savings goal (e.g., holiday, car, home deposit)? ")

# New feature 1- ask for monthly saving amount
monthly_saving_amount = float(input("How much do you plan to save each month? "))

# New feature 2- ask for total goal amount
goal_amount = float(input("What is the total amount you need for your savings goal? "))

# Calculate yearly savings
yearly_saving_total = monthly_saving_amount * 12

# Calculate months needed to reach the goal
months_to_goal = goal_amount / monthly_saving_amount

years_to_goal = months_to_goal / 12

# Build the personalised welcome message
welcome_message = (
    f"Welcome to your Personal Finance Tracker, {user_name}! "
    f"Your profile is now set up. The app will use a {theme_color} theme, "
    f"track all amounts in {primary_currency}, and organise your income on a {income_frequency} schedule. "
    f"Your monthly budget limit is set to {monthly_budget_limit}. "
    f"Your main savings goal is: {main_savings_goal}. "
)

# Print the final message and the new feature results
print("\n--- Finance Tracker Profile Summary ---")
print(welcome_message)
print(f"If you save {monthly_saving_amount} each month, you will save {yearly_saving_total} in a year.")
print(f"To reach your goal of {goal_amount}, it will take approximately {months_to_goal:.1f} months.")
print(f"That is about {years_to_goal:.2f} years.")

#side note .2f or .1f is a formatting instruction used inside an f‑string.
#It tells Python: “Show this number with 2 decimal places, even if it doesn’t need them.”
````
