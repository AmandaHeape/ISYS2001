## The three functions your code has to get right

Build your dictionary however you like, then write these three functions to work on it.
The checks at the bottom call them by name, so use these names and have them behave as
described:

- `total_for(spending, category)` returns the amount spent in that category, and returns
  `0` for a category that was never recorded. It must not crash on a missing category.
- `add_transaction(spending, transaction)` takes one transaction, in the same shape as
  the records in the list above, and updates your dictionary so that amount is counted
  from then on.
- `summary(spending)` returns a dictionary that maps every category to its total.

How `spending` is built and shaped inside is entirely your decision. The checks only care
about what these three functions do, not how you store things underneath.

```python
# Build your dictionary from the transactions, and write your three functions here.

spending = {}
for transaction in transactions:
    category = transaction['category']
    amount = transaction['amount']
    spending[category] = spending.get(category, 0) + amount

# Display the initial spending dictionary
print("Initial spending dictionary:")
print(spending)

# 1. total_for(spending, category)
def total_for(spending_data, category):
    """
    Returns the total amount spent in a given category.
    Returns 0 if the category is not found.
    """
    return spending_data.get(category, 0)

# Example usage of total_for
print(f"\nTotal for groceries: {total_for(spending, 'groceries')}")
print(f"Total for rent: {total_for(spending, 'rent')}")
print(f"Total for a non-existent category: {total_for(spending, 'books')}")

# 2. add_transaction(spending, transaction)
def add_transaction(spending_data, transaction):
    """
    Adds a single transaction to the spending dictionary.
    Updates the total for the corresponding category.
    """
    category = transaction['category']
    amount = transaction['amount']
    spending_data[category] = spending_data.get(category, 0) + amount

# Example usage of add_transaction
print(f"\nSpending before adding new transaction: {spending}")
new_transaction = {'date': '2026-08-29', 'category': 'transport', 'amount': 20, 'description': 'Bus ticket'}
add_transaction(spending, new_transaction)
print(f"Spending after adding new transaction (transport +20): {spending}")

# 3. summary(spending)
def summary(spending_data):
    """
    Returns a dictionary that maps every category to its total.
    In this implementation, it simply returns the spending_data itself
    since it already holds category totals.
    """
    return spending_data

# Example usage of summary
print(f"\nSummary of spending: {summary(spending)}")
```

```python
# Your own feature goes here.
from datetime import datetime

def total_for_week(transactions_list, start_date_str, end_date_str):
    """
    Calculates the total spending for transactions within a specified week.
    start_date_str and end_date_str should be in 'YYYY-MM-DD' format.
    """
    start_date = datetime.strptime(start_date_str, '%Y-%m-%d')
    end_date = datetime.strptime(end_date_str, '%Y-%m-%d')

    weekly_total = 0
    for transaction in transactions_list:
        transaction_date = datetime.strptime(transaction['date'], '%Y-%m-%d')
        if start_date <= transaction_date <= end_date:
            weekly_total += transaction['amount']
    return weekly_total

# Example usage:
# Let's find the total spending for the first week of August (Aug 1 to Aug 7)
week_start = '2026-08-01'
week_end = '2026-08-07'
weekly_spending = total_for_week(transactions, week_start, week_end)
print(f"Total spending from {week_start} to {week_end}: ${weekly_spending}")

# Another example: a week with no transactions (e.g., in September)
week_start_sept = '2026-09-01'
week_end_sept = '2026-09-07'
weekly_spending_sept = total_for_week(transactions, week_start_sept, week_end_sept)
print(f"Total spending from {week_start_sept} to {week_end_sept}: ${weekly_spending_sept}")
```
